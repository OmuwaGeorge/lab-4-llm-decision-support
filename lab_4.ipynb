# Section 1 — Talking to an LLM Programmatically

Before building anything, understand the anatomy of an API call: **messages and roles**
(`system`, `user`, `assistant`), and the **generation parameters** (`temperature`,
`max_tokens`).

### Part 1.1 — Your first API call

````python
# TODO: Write a helper function you will reuse for the WHOLE lab:
#
def ask_llm(user_prompt, system_prompt="You are a helpful assistant.",
    temperature=0.7, max_tokens=500):
  response = client.chat.completions.create(
  model=MODEL,
  messages=[
      {"role": "system", "content": system_prompt},
      {"role": "user",   "content": user_prompt},
  ],
  temperature=temperature,
  max_tokens=max_tokens,
  )
  return response.choices[0].message.content
#
# TODO: Call it once with a simple question and print the answer.
question = ask_llm("What is your purpose?")
# TODO: Print response.usage as well — how many tokens did your call consume?
print(question)

response = client.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "What is your purpose?"},
    ],
    temperature=0.7,
    max_tokens=500,
)

print(response.choices[0].message.content)
print(response.usage)

# My call consumed 229 tokens
````

> **Answer:**

1. The system role handles the models abilities and restrictions, basically what the model can and cannot do. The user role has to do with the actual request currently being made.

2. From what we looked at in class, a token can basically be a word, subword or character.
Billing per token makes more sense for all cases as someone might make 10 seperate requests that are all short and easy for the model to assess, but another person may make one request with over a 1000 words which will take a lot of computing power.
